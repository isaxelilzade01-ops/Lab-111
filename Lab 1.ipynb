#Avtobuslar

```python
N = int(input("How many students are there? "))
Yer = 40

if N <= Yer:
    print(f"Number of free Spaces: {Yer-N}, Number of Busses: 1")

elif N > Yer:
    C = N // Yer
    D = Yer - (N % Yer)

    print(f"Number of free spaces: {D}, Number of Busses: {C + 1}")
```
